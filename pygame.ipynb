Introducción: ¿Cómo piensa un videojuego?
Antes de escribir código, es clave enteder que un videojuego no es como un script de consola que corre y termina. Un videojuego es un bucle infinito que se ejecuta a gran velocidad (FPS), repitiendo tres pasos continuamente.  



1.   Leer entradas (teclado, mouse, joystick).
2.   Actualizar la lógica (mover personajes, calcular colisiones, cambiar puntajes).
3.   EDibujar en pantalla (borrar el cuadro anterior y pintar el nuevo).

**import pygame:** Trae la librería con todas las funciones gráficas y de sonido.

**import sys:** Se explica más adelante, pero sirve para interactuar directamente con el sistema operativo (cerrar la ventana de forma limpia).

**pygame.init()**: Este comando es obligatorio. Despierta internamente los módulos de audio, video y fuentes de pygame. Si no se pone, las funciones de dibujo o sonido fallarán.

```python
import pygame
import sys

# Inicializar todos los módulos de pygame
pygame.init()
```

Paso 2: Configuración de la Pantalla (El Lienzo)
Definimos el tamaño de la ventana del juego. Aquí repasamos el concepto de tuplas y coordenadas fijas.

**pygame.display.set_mode((ANCHO, ALTO))**: Abre la ventana. Es crucial notar que recibe una tupla (ANCHO, ALTO) como argumento (doble paréntesis).

El sistema de coordenadas: Explícales que en computación el origen (0,0) está en la esquina superior izquierda.

Si aumentas X, vas a la derecha.

Si aumentas Y, vas hacia abajo (a diferencia del plano cartesiano matemático).

**pygame.display.set_caption():** Modifica el texto que aparece en la barra superior de la ventana.

```python
# Definir dimensiones de la pantalla (Ancho, Alto)
ANCHO = 800
ALTO = 600

# Crear la ventana
pantalla = pygame.display.set_mode((ANCHO, ALTO))
pygame.display.set_caption("Mi Primer Videojuego")
```

Paso 3: El Reloj y los FPS (Frames per Second)
Para que el juego no corra a velocidades distintas en computadoras más rápidas o lentas, controlamos el tiempo.

```python
reloj = pygame.time.Clock()
```

**pygame.time.Clock():** Crea un objeto que nos servirá para limitar cuántas veces por segundo se actualiza el juego.

Paso 4: El Bucle Principal y la Captura de Eventos
Aquí es donde aplican el while y el for. El juego necesita quedarse escuchando lo que hace el usuario.

```python
# Bandera para controlar el bucle
jugando = True

while jugando:
    # 1. CAPTURA DE EVENTOS (Teclado, Mouse, cerrar ventana)
    for evento in pygame.event.get():
        if evento.type == pygame.QUIT:
            jugando = False
```

**pygame.event.get(): **Es una lista de todo lo que pasó en la computadora en esa milésima de segundo (clics, teclas presionadas, mover el mouse).

El bucle for: Revisa uno por uno esos eventos.

**evento.type == pygame.QUIT**: Detecta si el usuario hizo clic en la "X" de la ventana. Al cambiar jugando = False, el while termina.

Paso 5: Captura del Teclado para el Comportamiento (Movimiento)
Para mover un objeto, necesitamos capturar el estado continuo del teclado y asociarlo a variables de posición (x, y).

Nota para la clase: Antes del while, define variables como jugador_x = 400, jugador_y = 500 y velocidad = 5.

Dentro del while (después del bucle de eventos), agregamos:

```python
# 2. CAPTURA CONTINUA DEL TECLADO
    teclas = pygame.key.get_pressed()

    if teclas[pygame.K_LEFT]:
        jugador_x -= velocidad
    if teclas[pygame.K_RIGHT]:
        jugador_x += velocidad
    if teclas[pygame.K_UP]:
        jugador_y -= velocidad
    if teclas[pygame.K_DOWN]:
        jugador_y += velocidad
```

**pygame.key.get_pressed():** Devuelve una lista de estados (Verdadero/Falso) para todas las teclas del teclado. Es ideal para movimiento fluido (dejar la tecla presionada).

**pygame.K_LEFT, pygame.K_RIGHT, etc.: **Son las constantes que representan las flechas del teclado (puedes usar pygame.K_a, pygame.K_d si prefieren WASD).

Modificación de coordenadas: Si presionan izquierda, restamos a X. Si presionan abajo, sumamos a Y.

Paso 6: Dibujo y Renderizado
Una vez que las variables cambiaron de valor, hay que plasmarlas en la pantalla.

```python
# 3. DIBUJO
    # Limpiar la pantalla con un color (R, G, B)
    pantalla.fill((0, 0, 0))  # Negro

    # Dibujar un personaje básico (un rectángulo)
    # pygame.draw.rect(superficie, color, (x, y, ancho, alto))
    pygame.draw.rect(pantalla, (255, 0, 0), (jugador_x, jugador_y, 50, 50))

    # Actualizar la pantalla completa
    pygame.display.flip()

    # Controlar los FPS
    reloj.tick(60)
```

**pantalla.fill((0, 0, 0)):** Si no limpiamos la pantalla en cada frame, el personaje dejará un "rastro" o línea pintada por donde pase. Usamos formato RGB (Rojo, Verde, Azul).

**pygame.draw.rect():** Función nativa para dibujar rectángulos. Pide dónde dibujarlo (pantalla), de qué color (Tupla RGB, ej: (255, 0, 0) es Rojo) y un rectángulo de posición y tamaño (x, y, ancho, alto).

**pygame.display.flip(): **Es el comando que realmente "vuelca" todo lo que dibujamos en la memoria RAM hacia el monitor. Sin esto, la pantalla se queda en negro.

**reloj.tick(60): **Detiene el bucle lo necesario para que el juego corra exactamente a 60 cuadros por segundo.

Paso 7: Cierre Limpio del Programa
Fuera del bucle while (cuando jugando pasa a ser False):

```python
# Salida del juego
pygame.quit()
sys.exit()
```

jnuego de muestra **Atrapa los objetos**

```python
import pygame
import sys
import random

# 1. INICIALIZACIÓN Y CONFIGURACIÓN
pygame.init()

ANCHO = 800
ALTO = 600
pantalla = pygame.display.set_mode((ANCHO, ALTO))
pygame.display.set_caption("¡Atrapa las Estrellas! ")
reloj = pygame.time.Clock()

# Colores (RGB)
NEGRO = (15, 15, 25)
BLANCO = (255, 255, 255)
VERDE = (50, 205, 50)
AMARILLO = (255, 215, 0)
ROJO = (220, 20, 60)

# 2. VARIABLES DEL JUGADOR (El cubo verde)
jugador_ancho = 60
jugador_alto = 20
jugador_x = ANCHO // 2 - jugador_ancho // 2
jugador_y = ALTO - 50
velocidad_jugador = 8

# 3. VARIABLES DELOBJETIVO (El cuadrado amarillo que cae)
objetivo_tamano = 30
objetivo_x = random.randint(0, ANCHO - objetivo_tamano)
objetivo_y = -objetivo_tamano
velocidad_objetivo = 5

# 4. VARIABLES DEL JUEGO
puntuacion = 0
vidas = 3
fuente = pygame.font.SysFont("monospace", 24) # Fuente para el texto

# 5. BUCLE PRINCIPAL DEL JUEGO
jugando = True
game_over = False

while jugando:

    # --- GESTIÓN DE EVENTOS ---
    for evento in pygame.event.get():
        if evento.type == pygame.QUIT:
            jugando = False

        # Reiniciar juego si perdieron y presionan la barra espaciadora
        if game_over and evento.type == pygame.KEYDOWN:
            if evento.key == pygame.K_SPACE:
                puntuacion = 0
                vidas = 3
                velocidad_objetivo = 5
                objetivo_y = -objetivo_tamano
                game_over = False

    if not game_over:
        # --- LÓGICA Y CAPTURA DE TECLADO ---
        teclas = pygame.key.get_pressed()

        if teclas[pygame.K_LEFT] and jugador_x > 0:
            jugador_x -= velocidad_jugador
        if teclas[pygame.K_RIGHT] and jugador_x < ANCHO - jugador_ancho:
            jugador_x += velocidad_jugador

        # Movimiento automático del objeto hacia abajo
        objetivo_y += velocidad_objetivo

        # ¿El objeto tocó el suelo? (El jugador no lo atrapó)
        if objetivo_y > ALTO:
            objetivo_x = random.randint(0, ANCHO - objetivo_tamano)
            objetivo_y = -objetivo_tamano
            vidas -= 1
            if vidas <= 0:
                game_over = True

        # DETECCIÓN DE COLISIÓN (¿El jugador atrapó el objeto?)
        # Verificamos si los rectángulos se superponen
        if (jugador_x < objetivo_x + objetivo_tamano and
            jugador_x + jugador_ancho > objetivo_x and
            jugador_y < objetivo_y + objetivo_tamano and
            jugador_y + jugador_alto > objetivo_y):

            puntuacion += 1
            objetivo_x = random.randint(0, ANCHO - objetivo_tamano)
            objetivo_y = -objetivo_tamano

            # Aumentar la dificultad poco a poco
            if puntuacion % 3 == 0:
                velocidad_objetivo += 1

    # --- DIBUJO EN PANTALLA ---
    pantalla.fill(NEGRO)

    if not game_over:
        # Dibujar Jugador
        pygame.draw.rect(pantalla, VERDE, (jugador_x, jugador_y, jugador_ancho, jugador_alto))

        # Dibujar Objetivo
        pygame.draw.rect(pantalla, AMARILLO, (objetivo_x, objetivo_y, objetivo_tamano, objetivo_tamano))

        # Mostrar Puntuación y Vidas
        texto_puntos = fuente.render(f"Puntos: {puntuacion}", True, BLANCO)
        texto_vidas = fuente.render(f"Vidas: {vidas}", True, ROJO)
        pantalla.blit(texto_puntos, (10, 10))
        pantalla.blit(texto_vidas, (ANCHO - 150, 10))
    else:
        # Pantalla de Game Over
        texto_game_over = fuente.render("¡GAME OVER!", True, ROJO)
        texto_reiniciar = fuente.render("Presiona ESPACIO para volver a jugar", True, BLANCO)

        # Centrar los textos en pantalla
        pantalla.blit(texto_game_over, (ANCHO // 2 - 80, ALTO // 2 - 20))
        pantalla.blit(texto_reiniciar, (ANCHO // 2 - 240, ALTO // 2 + 20))

    # Actualizar pantalla y reloj
    pygame.display.flip()
    reloj.tick(60)

# Cierre ordenado
pygame.quit()
sys.exit()
```
