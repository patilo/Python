 ## 1. Rellenar Listas con Números

 ### Definición de Lista
 Una **lista** en Python es una colección ordenada y mutable de elementos. Puede contener elementos de diferentes tipos de datos (enteros, flotantes, cadenas, etc.). Se definen usando corchetes `[]` y los elementos están separados por comas.

 ### ¿Por qué usar un bucle `for`?
 Cuando necesitas crear una lista con una secuencia de números o aplicar una operación a cada número en un rango para añadirlo a la lista, un bucle `for` es la herramienta perfecta. Evita escribir manualmente cada elemento, especialmente para listas grandes.

 ### Ejemplo 1: Lista de números secuenciales

```python
 # Ejemplo: Crear una lista de números del 0 al 4
 print("Ejemplo 1: Lista de números secuenciales")
 sequential_numbers = []
 for i in range(5): # Rango de 0 a 4
   sequential_numbers.append(i) #aqui usamos la funcion append para ingresarlos a la lista (i)
 print(f"Lista de números secuenciales: {sequential_numbers}")
```

 ### Ejemplo 2: Lista de cuadrados de números

```python
 # Ejemplo: Crear una lista de los cuadrados de los números del 1 al 5
 print("\nEjemplo 2: Lista de cuadrados de números")
 squares = []
 for i in range(1, 6): # Números del 1 al 5
   squares.append(i * i) #como i almacena el valor lo multiplica por el mismo simulando una potencia
 print(f"Lista de cuadrados: {squares}")
```

 ### Ejercicio Resuelto 1: Generar números pares

 **Problema:** Crea una lista que contenga todos los números pares del 0 al 10 .

 **Solución:**

```python
 # Ejercicio Resuelto 1: Números pares del 0 al 10
 print("\nEjercicio Resuelto 1: Números pares del 0 al 10")
 even_numbers = []
 for i in range(0, 11, 2): # Números pares del 0 al 10 (0, 2, 4, 6, 8, 10)
   even_numbers.append(i)
 print(f"Números pares: {even_numbers}")
```

 ## 2. Rellenar Listas con Cadenas de Texto (Strings)

 ### ¿Por qué usar un bucle `for`?
 Similar a los números, un bucle `for` es ideal para generar listas de cadenas cuando necesitas crear nombres secuenciales, formatear texto o procesar una colección de elementos para convertirlos en cadenas con un patrón específico.

 ### Ejemplo 3: Lista de nombres formateados

```python
 # Ejemplo: Crear una lista de nombres de productos
 print("\nEjemplo 3: Lista de nombres formateados")
 product_names = []
 for i in range(1, 6):
   product_names.append(f"Producto_{i}")
 print(f"Nombres de productos: {product_names}")
```

 ### Ejercicio Resuelto 2: Crear una lista de estados

```python
 # Ejercicio Resuelto 2: Lista de estados basada en una lista inicial
 print("\nEjercicio Resuelto 2: Lista de estados")
 raw_statuses = ["pending", "approved", "rejected"]
 formatted_statuses = []
 for status in raw_statuses:
   formatted_statuses.append(f"Status: {status}!")
 print(f"Estados formateados: {formatted_statuses}")
```

 ## 3. Rellenar Diccionarios con Claves y Valores

 ### Definición de Diccionario
 Un **diccionario** en Python es una colección desordenada, mutable y indexada de elementos. Almacena datos en pares **clave-valor**. Cada clave debe ser única y se utiliza para acceder a su valor asociado. Se definen usando llaves `{}`.

 ### ¿Por qué usar un bucle `for`?
 Los bucles `for` son excepcionalmente útiles para construir diccionarios dinámicamente, ya sea emparejando dos listas (una para claves y otra para valores), o generando claves y valores a partir de una única fuente de datos.

 ### Ejemplo 4: Diccionario de número a su cuadrado

```python
 # Ejemplo: Crear un diccionario donde la clave es un número y el valor es su cuadrado
 print("\nEjemplo 4: Diccionario de número a su cuadrado")
 squares_dict = {}
 for i in range(1, 6): # Números del 1 al 5
   squares_dict[i] = i * i
 print(f"Diccionario de cuadrados: {squares_dict}")
```

 ### Ejemplo 5: Diccionario a partir de una lista de tuplas

```python
 # Ejemplo: Crear un diccionario a partir de una lista de tuplas (nombre, edad)
 print("\nEjemplo 5: Diccionario a partir de una lista de tuplas")
 people_data = [("Ana", 25), ("Luis", 30), ("Marta", 22)]
 people_dict = {}
 for item in people_data:
   key = item[0]
   value = item[1]
   people_dict[key] = value
 print(f"Diccionario de personas: {people_dict}")
```

 ### Ejercicio Resuelto 3: Conteo de elementos en una lista

 **Problema:** Dada una lista de frutas, crea un diccionario que cuente cuántas veces aparece cada fruta.

 **Solución:**

```python
 # Ejercicio Resuelto 3: Conteo de elementos en una lista
 print("\nEjercicio Resuelto 3: Conteo de frutas")
 fruits = ["manzana", "banana", "manzana", "naranja", "banana", "manzana"]

 # Una forma de hacerlo es inicializar el diccionario y luego iterar
 fruit_counts = {}
 for fruit in fruits:
   if fruit in fruit_counts:
     fruit_counts[fruit] += 1
   else:
     fruit_counts[fruit] = 1

 print(f"Conteo de frutas: {fruit_counts}")

 # Otra forma, más concisa con collections.Counter (para conocimiento)
 from collections import Counter
 fruit_counts_counter = Counter(fruits)
 print(f"Conteo de frutas (con Counter): {fruit_counts_counter}")
```

Vamos a programar el módulo de carga para un sistema de monitoreo ambiental en la región de Tarapacá. El usuario debe indicar cuántas estaciones meteorológicas registrará hoy, y luego el sistema generará el diccionario correspondiente.

```python
print("--- CONFIGURACIÓN INICIAL DEL SISTEMA ---")

# 1. BUCLE WHILE: Validación de entrada para asegurar un número válido
cantidad_estaciones = 0

while cantidad_estaciones <= 0:
    cantidad_estaciones = int(input("¿Cuántas estaciones desea registrar hoy? (Mínimo 1): "))

    if cantidad_estaciones <= 0:
        print(" Error: Debe ingresar una cantidad mayor a cero.\n")

print(f"\n Configuración aceptada: Se registrarán {cantidad_estaciones} estaciones.")

# 2. BUCLE FOR CON RANGE: Usamos la cantidad validada para el llenado
catastro_ambiental = {}

# El rango dará exactamente las vueltas que el usuario solicitó en el while
for i in range(1, cantidad_estaciones + 1):
    print(f"\n Datos de la Estación N°{i}:")

    # Solicitamos los datos que compondrán nuestra estructura Clave:Valor
    nombre_zona = input("Ingrese el nombre de la comuna/zona (Clave): ").strip().capitalize()
    temperatura = float(input(f"Ingrese la temperatura actual en {nombre_zona} (°C): "))

    # 3. LLENADO DEL DICCIONARIO: Asignación directa
    # La clave es el string de la zona y el valor es el número decimal de la temperatura
    catastro_ambiental[nombre_zona] = round(temperatura, 1)


# 4. REPORTE FINAL DE PRODUCCIÓN
print(" REPORTE DE DATOS AMBIENTALES CAPTURADOS")
print(f"Total de registros en diccionario: {len(catastro_ambiental)}")
print(f"Diccionario completo: {catastro_ambiental}")
```
