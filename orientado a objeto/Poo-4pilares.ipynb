**Encapsulamiento, Getters y Setters**

Imagina un Cajero Automático (ATM):

El cajero tiene dentro tu dinero (un dato guardado en una variable saldo).

Si el cajero dejara que le abras la puerta y agarres el billete que quieras, sería un caos (acceso directo a la variable).

En su lugar, el cajero oculta la puerta (encapsulamiento) y te da una pantalla táctil con botones específicos para sacar o meter dinero.

**Encapsulamiento** es ocultar los datos internos de un objeto para que nadie los altere por error.

Si dejamos las variables públicas, cualquiera puede modificarlas desde fuera sin control, rompiendo la lógica del programa (por ejemplo, poner vida negativa o un salario de -$5,000).

**Getter** (Obtener): Un método para leer un dato protegido.

**Setter** (Establecer/Modificar): Un método para cambiar un dato protegido pasando por reglas de validación.

Fundamento: Los objetos vacíos no sirven de mucho. Necesitamos que cada objeto nazca con sus propias características (atributos) y sepa hacer cosas (métodos).

Explicación
__init__ (Constructor): Es la función especial que se ejecuta automáticamente cuando creamos un objeto. Sirve para darle sus datos iniciales.

self: Representa al objeto mismo que se está creando. Dice: "este nombre pertenece a ESTE personaje en específico, no a otro".

Métodos: Son funciones que viven dentro de la clase y definen qué puede hacer el objeto.

```python
class Personaje:
    # Constructor: Nace con nombre y vida
    def __init__(self, nombre, vida):
        self.nombre = nombre  # Atributo
        self.vida = vida      # Atributo

    # Método: Comportamiento del personaje
    def atacar(self, enemigo):
        print(f"{self.nombre} ataca a {enemigo}!")

# Creamos los personajes con sus datos iniciales
guerrero = Personaje("Aragorn", 100)
mago = Personaje("Gandalf", 80)

# Usamos sus atributos y métodos
print(f"Vida de {guerrero.nombre}: {guerrero.vida}")
mago.atacar("Orco")
```

Con Encapsulamiento

*Nota: En Python se usa un guion
bajo _ o doble guion __ para indicar que una variable es privada.*

```python
class CuentaBancaria:
    def __init__(self, titular, saldo_inicial):
        self.titular = titular
        self.__saldo = saldo_inicial  # Variable PRIVADA (oculta al exterior)

    # GETTER: Permite ver el saldo
    def obtener_saldo(self):
        return self.__saldo

    # SETTER: Permite cambiar el saldo, PERO con validación
    def depositar(self, monto):
        if monto > 0:
            self.__saldo += monto
            print(f"Depósito exitoso. Nuevo saldo: ${self.__saldo}")
        else:
            print("Error: No puedes depositar un monto negativo o cero.")

# USO SEGURO
cuenta = CuentaBancaria("Carlos", 1000)

# print(cuenta.__saldo)  #Da error porque es privada
print(cuenta.obtener_saldo())  #Salida: 1000 (Usamos el Getter)

cuenta.depositar(-500)  #Salida: Error: No puedes depositar un monto negativo
cuenta.depositar(500)   #Salida: Depósito exitoso. Nuevo saldo: $1500
```

Herencia:
Imagínate la relación entre Smartphone y Teléfono Antiguo:

Ambos tienen pantalla/auricular y ambos pueden hacer llamadas.

El Smartphone "hereda" las capacidades del teléfono tradicional, pero añade cámara, aplicaciones e internet.

En código, la Herencia permite crear una clase "Padre" con características generales y clases "Hijas" que heredan todo lo del padre y añaden lo suyo propio. Nos ahorra duplicar código.

```python
# CLASE PADRE (Base)
class Vehiculo:
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo

    def encender(self):
        print(f"El {self.marca} {self.modelo} está encendido.")
```

```python
CLASE HIJA 1 (Hereda de Vehiculo)
class Auto(Vehiculo):
    def __init__(self, marca, modelo, num_puertas):
        # Invocamos al constructor del Padre (super)
        super().__init__(marca, modelo)
        self.num_puertas = num_puertas  # Propiedad única de Auto

    def tocar_bocina(self):
        print("pip pip")
```

```python
CLASE HIJA 2 (Hereda de Vehiculo)
class Moto(Vehiculo):
    def hacer_caballito(self):
        print("¡Levantando la rueda delantera!")
```

```python
# USO
mi_carro = Auto("Toyota", "Corolla", 4)
mi_carro.encender()       # Heredado del Padre: El Toyota Corolla está encendido.
mi_carro.tocar_bocina()   # Propio de la Hija Auto.

mi_moto = Moto("Yamaha", "MT-07")
mi_moto.encender()        # Heredado del Padre.
mi_moto.hacer_caballito() # Propio de Moto.
```

OTRO EJEMPLO DE HERENCIA

```python
# CLASE PADRE
class Personaje:
    def __init__(self, nombre, vida):
        self.nombre = nombre
        self.__vida = vida

    def get_vida(self):
        return self.__vida

    def recibir_daño(self, daño):
        self.__vida -= daño
        print(f"{self.nombre} recibió {daño} de daño. Vida restante: {self.__vida}")

# CLASE HIJA 1
class Mago(Personaje):  # Mago hereda de Personaje
    def __init__(self, nombre, vida, mana):
        super().__init__(nombre, vida)  # Inicializa nombre y vida en el padre
        self.mana = mana               # Atributo exclusivo del Mago

    def lanzar_hechizo():
        print(f"✨ {self.nombre} lanza una bola de fuego usando {self.mana} de maná!")

# Uso
mi_mago = Mago("Gandalv", 80, 100)

# Métodos heredados del Padre
mi_mago.recibir_daño(20)

# Métodos propios de la Hija
mi_mago.lanzar_hechizo()
```

**POLIMORFISMO**

Qué significa?
La palabra viene del griego:

Poli = Muchos

Morfismo = Formas

En programación, el Polimorfismo es la capacidad que tienen objetos de diferentes clases de responder al mismo mensaje (o método), pero cada uno a su propia manera.

image.png

El polimorfismo se logra principalmente mediante la Sobrescritura de Métodos (Method Overriding):

La clase Padre define un método general.

Las clases Hijas rediseñan (sobrescriben) ese método con su propio comportamiento.

```python
# 1. CLASE PADRE
class Animal:
    def __init__(self, nombre):
        self.nombre = nombre

    def hacer_sonido(self):
        # Este método existe para ser sobrescrito por los hijos
        pass

# 2. CLASES HIJAS (Cada una implementa su propio sonido)
class Perro(Animal):
    def hacer_sonido(self):
        return "¡Guau, guau!"

class Gato(Animal):
    def hacer_sonido(self):
        return "¡Miau, miau!"

class Pato(Animal):
    def hacer_sonido(self):
        return "¡Cuac, cuac!"


# 3. DEMOSTRACIÓN DE POLIMORFISMO
# Creamos una lista con diferentes animales
granja = [
    Perro("Firulais"),
    Gato("Michi"),
    Pato("Lucas")
]

# Recorremos la lista y llamamos al MISMO método para todos
for animal in granja:
    # No nos importa qué tipo exacto de animal es,
    # sabemos que TODOS responden al método 'hacer_sonido()'
    print(f"{animal.nombre} dice: {animal.hacer_sonido()}")
```

¿Por qué es tan poderoso? (El Gran Beneficio)
Sin polimorfismo, tendrías que escribir código lleno de condicionales feos como este:

```python
# CÓDIGO MALO (Sin polimorfismo)
for animal in granja:
    if isinstance(animal, Perro):
        print(animal.ladrar())
    elif isinstance(animal, Gato):
        print(animal.maullar())
    elif isinstance(animal, Pato):
        print(animal.graznar())
```

Ejercicio Práctico Resuelto: Sistema de Pagos
Diseñaremos un procesador de pagos para un carrito de compras.
Crea una clase base MetodoPago con un método procesar_pago(monto).  
Crea tres clases hijas:

*TarjetaCredito:* Procesa el pago y cobra una comisión del 3%

*PayPal:* Procesa el pago solicitando un correo electrónico.

*Efectivo:* Aplica un descuento del $10% por pago al contado.

**Crea una función efectuar_compra(metodo_pago, monto) que aplique el polimorfismo.**

```python
# ==========================================
# 1. CLASE BASE (PADRE)
# ==========================================
class MetodoPago:
    def procesar_pago(self, monto):
        raise NotImplementedError("Las clases hijas deben implementar este método")
#raise NotImplementedError es una forma de obligar a las clases hijas a implementar un método.
#raise se usa para provocar intensionalmente y detener la ejecucion
#notImplementedError es un tipo de error que capta python que indica que existe pero no tiene code
#si no tuviera el raise, se deberia usar el pass y no avisaria ningun error
#por ende el raise obliga que la clase hija deba definir el metodo

# ==========================================
# 2. CLASES HIJAS (SOBRESCRITURA POLIMÓRFICA)
# ==========================================
class TarjetaCredito(MetodoPago):
    def __init__(self, numero_tarjeta):
        self.numero_tarjeta = numero_tarjeta

    def procesar_pago(self, monto):
        comision = monto * 0.03
        total = monto + comision
        print(f"Pago con Tarjeta {self.numero_tarjeta[-4:]}: Monto ${monto} + Comisión ${comision:.2f} = Total: ${total:.2f}")


class PayPal(MetodoPago):
    def __init__(self, email):
        self.email = email

    def procesar_pago(self, monto):
        print(f"Pago procesado vía PayPal ({self.email}): Total: ${monto:.2f}")


class Efectivo(MetodoPago):
    def procesar_pago(self, monto):
        descuento = monto * 0.10
        total = monto - descuento
        print(f"Pago en Efectivo: Monto ${monto} - Descuento 10% (${descuento:.2f}) = Total: ${total:.2f}")


# ==========================================
# 3. FUNCIÓN POLIMÓRFICA EN ACCIÓN
# ==========================================
def finalizar_compra(medio_de_pago, total_carrito):
    print("\nIniciando transacción...")
    # Polimorfismo: No importa qué medio de pago sea, todos ejecutan procesar_pago()
    medio_de_pago.procesar_pago(total_carrito)


# ==========================================
# 4. PRUEBAS DEL SISTEMA
# ==========================================
monto_compra = 100.0

# Instanciamos los diferentes métodos de pago
visa = TarjetaCredito("4532-1234-5678-9999")
cuenta_paypal = PayPal("usuario@email.com")
pago_efectivo = Efectivo()

# Probamos la misma función con 3 objetos totalmente distintos
finalizar_compra(visa, monto_compra)
finalizar_compra(cuenta_paypal, monto_compra)
finalizar_compra(pago_efectivo, monto_compra)
```
