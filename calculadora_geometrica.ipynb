```python
# =====================================================================
# SISTEMA DE CÁLCULO GEOMÉTRICO MODULAR
# =====================================================================

# Importamos el módulo math para usar la constante math.pi de alta precisión
import math

# ---------------------------------------------------------------------
# 1. BLOQUE DE FUNCIONES (Las "Máquinas" de Cálculo)
# ---------------------------------------------------------------------

# --- CUADRADO ---
def calcular_cuadrado(lado):
    perimetro = lado * 4
    area = lado ** 2
    volumen = lado ** 3 # Cubo (extensión tridimensional del cuadrado)
    return perimetro, area, volumen

# --- CÍRCULO / ESFERA ---
def calcular_circulo(radio):
    perimetro = 2 * math.pi * radio # También llamado longitud de circunferencia
    area = math.pi * (radio ** 2)
    volumen = (4/3) * math.pi * (radio ** 3) # Volumen de una esfera basada en ese radio
    # Retornamos los valores redondeados a 2 decimales para limpieza visual
    return round(perimetro, 2), round(area, 2), round(volumen, 2)

# --- TRIÁNGULO EQUILÁTERO / PRISMA ---
def calcular_triangulo(base, altura, lado):
    perimetro = lado * 3 # Asumiendo un triángulo equilátero para simplificar
    area = (base * altura) / 2
    # Simulamos el volumen de un prisma triangular recto con altura del prisma igual a la base
    volumen = area * base
    return perimetro, area, round(volumen, 2)


# ---------------------------------------------------------------------
# 2. PROGRAMA PRINCIPAL (Mesa de Control e Inputs)
# ---------------------------------------------------------------------

print("========================================")
print("📐 CALCULADORA GEOMÉTRICA MODULAR v1.0")
print("========================================")
print("1. Cuadrado (Cubo)")
print("2. Círculo (Esfera)")
print("3. Triángulo (Prisma)")
print("4. Salir")
print("----------------------------------------")

opcion = int(input("Seleccione la figura que desea calcular (1-4): "))

# Utilizamos match-case para estructurar las opciones del menú de forma limpia
match opcion:

    case 1:
        print("\n⏹CONFIGURACIÓN DEL CUADRADO")
        val_lado = float(input("Ingrese la medida del lado (cm): "))

        # Invocamos la función y 'desempaquetamos' los 3 valores que nos devuelve
        p, a, v = calcular_cuadrado(val_lado)

        print(f"\nResultados:")
        print(f"-> Perímetro: {p} cm")
        print(f"-> Área: {a} cm²")
        print(f"-> Volumen del Cubo: {v} cm³")

    case 2:
        print("\nCONFIGURACIÓN DEL CÍRCULO / ESFERA")
        val_radio = float(input("Ingrese el radio de la figura (cm): "))

        # Invocamos la función pasando el radio como argumento
        p, a, v = calcular_circulo(val_radio)

        print(f"\n Resultados (Redondeados a 2 decimales):")
        print(f"-> Perímetro (Circunferencia): {p} cm")
        print(f"-> Área del Círculo: {a} cm²")
        print(f"-> Volumen de la Esfera: {v} cm³")

    case 3:
        print("\n CONFIGURACIÓN DEL TRIÁNGULO")
        # El triángulo requiere más ingredientes (parámetros) para sus fórmulas
        val_base = float(input("Ingrese la base del triángulo (cm): "))
        val_altura = float(input("Ingrese la altura del triángulo (cm): "))
        val_lado = float(input("Ingrese la medida de uno de sus lados (cm): "))

        # Enviamos los 3 datos en el orden exacto que los pide la definición
        p, a, v = calcular_triangulo(val_base, val_altura, val_lado)

        print(f" Resultados:")
        print(f"-> Perímetro: {p} cm")
        print(f"-> Área: {a} cm²")
        print(f"-> Volumen del Prisma: {v} cm³")

    case 4:
        print("\nSistema cerrado. ¡Gracias por usar la calculadora!")

    case _:
        print("\nError: Opción no válida en el sistema.")

print("\n========================================")
```
