## Ejercicio Grupal 1: Gestión de Tareas Pendientes (To-Do List)

**Enunciado:** Desarrollen un programa interactivo de línea de comandos que permita a un grupo de usuarios gestionar una lista de tareas pendientes. El programa debe:

1.  Mostrar un menú principal con las siguientes opciones:
    *   `1. Ver todas las tareas`
    *   `2. Añadir nueva tarea`
    *   `3. Marcar tarea como completada`
    *   `4. Eliminar tarea`
    *   `5. Salir`
2.  Utilizar una lista de diccionarios para almacenar las tareas. Cada diccionario debe tener al menos las claves `nombre` (str), `descripcion` (str), y `completada` (bool).
3.  Permitir añadir nuevas tareas, solicitando el nombre y la descripción.
4.  Permitir marcar una tarea como completada, identificándola por su índice o nombre.
5.  Permitir eliminar una tarea, identificándola por su índice o nombre.
6.  El menú debe repetirse hasta que el usuario elija la opción de salir.
7.  Asegurarse de que el programa maneje entradas inválidas del usuario (por ejemplo, opciones de menú incorrectas o índices fuera de rango).
8.  Al mostrar las tareas, indicar claramente cuáles están completadas y cuáles no.

## Ejercicio Grupal 2: Inventario de Tienda Simple

**Enunciado:** Construyan un sistema de inventario básico para una pequeña tienda. El programa debe ser interactivo y permitir al usuario realizar operaciones sobre los productos. El sistema debe:

1.  Mostrar un menú con las siguientes opciones:
    *   `1. Ver inventario`
    *   `2. Añadir producto`
    *   `3. Actualizar stock de producto`
    *   `4. Eliminar producto`
    *   `5. Buscar producto por nombre`
    *   `6. Salir`
2.  Utilizar un diccionario donde las claves sean los nombres de los productos (str) y los valores sean otros diccionarios que contengan `precio` (float) y `stock` (int).
3.  Permitir añadir nuevos productos, solicitando nombre, precio y stock inicial.
4.  Permitir actualizar el stock de un producto existente.
5.  Permitir eliminar un producto del inventario.
6.  Permitir buscar un producto por su nombre y mostrar toda su información.
7.  El menú debe persistir hasta que el usuario decida salir.
8.  Validar la entrada del usuario para asegurar que los precios sean números y el stock sea un entero positivo.

## Ejercicio Grupal 3: Sistema de Reserva de Asientos (Cine/Teatro)

**Enunciado:** Creen un programa para simular un sistema de reserva de asientos para una sala de cine o teatro. El objetivo es que los usuarios puedan ver el estado de los asientos y reservar los disponibles. El programa debe:

1.  Inicializar una sala de asientos como una lista de listas (o una lista de cadenas, simulando filas), donde cada elemento representa un asiento (`'O'` para libre, `'X'` para ocupado).
2.  Mostrar un menú con las siguientes opciones:
    *   `1. Ver asientos disponibles`
    *   `2. Reservar asiento`
    *   `3. Cancelar reserva`
    *   `4. Salir`
3.  Al ver los asientos, mostrar una representación clara de la sala.
4.  Permitir al usuario reservar un asiento, solicitando la fila y el número de asiento. Validar que el asiento esté disponible y que la entrada sea válida.
5.  Permitir al usuario cancelar una reserva, solicitando la fila y el número de asiento. Validar que el asiento esté ocupado y que la entrada sea válida.
6.  El programa debe seguir funcionando hasta que el usuario elija salir.
7.  Manejar posibles errores de entrada del usuario (ej. coordenadas inválidas, asientos ya ocupados/libres).

## Ejercicio Grupal 4: Registro de Estudiantes y Notas

**Enunciado:** Desarrollen un programa para gestionar un registro simple de estudiantes y sus notas en diferentes asignaturas. El programa debe ser interactivo y permitir:

1.  Mostrar un menú con las siguientes opciones:
    *   `1. Ver todos los estudiantes y sus notas`
    *   `2. Añadir nuevo estudiante`
    *   `3. Añadir/Actualizar nota de estudiante`
    *   `4. Calcular promedio de notas de un estudiante`
    *   `5. Salir`
2.  Utilizar un diccionario principal donde las claves sean los nombres de los estudiantes (str) y los valores sean diccionarios que contengan las asignaturas y sus respectivas notas (ej., `{'Matematicas': 85, 'Historia': 78}`).
3.  Permitir añadir nuevos estudiantes, inicialmente sin notas o con un diccionario de notas vacío.
4.  Permitir añadir o actualizar una nota para un estudiante específico en una asignatura dada.
5.  Permitir calcular y mostrar el promedio de notas de un estudiante, si tiene al menos una nota registrada.
6.  El menú debe estar dentro de un bucle `while` que termine cuando el usuario decida salir.
7.  Asegurar que todas las entradas de notas sean numéricas y que el estudiante exista antes de intentar añadir o actualizar notas.
