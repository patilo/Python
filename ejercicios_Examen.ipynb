## Ejercicio 1: Calculadora de Geometría 2D


##Desarrolla un programa en Python que funcione como una calculadora de geometría para figuras 2D. El programa debe presentar un menú al usuario para seleccionar entre 'Cuadrado', 'Círculo', 'Triángulo' y 'Rectángulo'. Una vez seleccionada la figura, el programa deberá solicitar al usuario las dimensiones necesarias. Finalmente, el programa debe calcular y mostrar el área y el perímetro de la figura seleccionada, incluyendo validación para que las entradas sean números positivos.


##Ejercicio 2: Conversor de Unidades (Longitud): Desarrolla un programa en Python que funcione como un conversor de unidades para medidas de longitud. El programa debe presentar un menú al usuario para seleccionar el tipo de conversión deseada (por ejemplo, 'Metros a Kilómetros', 'Pies a Metros', 'Millas a Kilómetros'). Una vez seleccionada la opción, el programa deberá solicitar al usuario el valor numérico a convertir. Finalmente, el programa debe realizar la conversión utilizando funciones apropiadas para cada tipo de cálculo y mostrar el resultado. Asegúrate de incluir validación para que las entradas sean números positivos.

  
##Ejercicio 3: Conversor de Unidades (Masa y Temperatura): Crea un programa en Python que extienda la funcionalidad de conversión de unidades para incluir medidas de masa y temperatura. El programa debe ofrecer un menú que permita al usuario elegir entre conversiones de masa (por ejemplo, 'Kilogramos a Libras', 'Gramos a Onzas') y conversiones de temperatura (por ejemplo, 'Celsius a Fahrenheit', 'Fahrenheit a Kelvin'). Después de seleccionar la opción, el programa debe pedir el valor a convertir. Deberá calcular y presentar el resultado utilizando funciones específicas para cada tipo de conversión. Implementa la validación para asegurar que los valores introducidos sean números válidos.
